```python
df = pd.read_csv('yahoo_stock.csv')

df
```

Preparation

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error

from statsmodels.tsa.arima.model import ARIMA
from pmdarima import auto_arima
```

```python
# Jika date masih kolom biasa
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])
    df = df.sort_values('Date')
    df = df.set_index('Date')
    
# Index berbentuk datetime
df.index = pd.to_datetime(df.index)

# Urutkan Index
df = df.sort_index()

# Set frequency businness day untuk data saham
df = df.asfreq('B')

# isi missing value karena weekend/libur bursa
df['Close'] = df['Close'].ffill()

series = df['Close']

print(series.head())
print(series.index.freq)
```

Train Test Split

```python
# 80% train
train_size = int(len(series) * 0.8)

train = series[:train_size]
test = series[train_size:]

print("Train 80%:", len(train))
print("Test 20%:", len(test))
```

Define best parameters

```python
# Mencari parameter terbaik (p, d, q)

auto_model = auto_arima(
    train,
    seasonal=False,  # Non-seasonal data
    stepwise=True,  # Stepwise search untuk efisiensi
    suppress_warnings=True,  # Suppress warnings
    trace=True,  # Trace hasil
)

print(auto_model.summary())
```

```python
# Model terbaik
order = auto_model.order

print("Best ARIMA order:", order)

model = ARIMA(train, order=order)

model_fit = model.fit()

print(model_fit.summary())
```

Prediksi

```python
predictions = model_fit.forecast(steps=len(test))

predictions.index = test.index

print(predictions.head())
```

Evaluasi

```python
rmse = np.sqrt(mean_squared_error(test, predictions))

mape = mean_absolute_percentage_error(test, predictions) * 100

mean_price = series.mean()
rmse_percent = (rmse / mean_price) * 100

print("MEAN :", mean_price)
print(f"RMSE: {rmse:.2f}")
print(f"MAPE: {mape:.2f}%")
```

Train ulang dengan full data

```python
final_model = ARIMA(series, order=order)

final_model_fit = final_model.fit()
```

prediksi 30 hari kedepan

```python
forecast_30 = final_model_fit.forecast(steps=30)

print(forecast_30)
```

Membuat tanggal prediksi

```python
future_dates = pd.date_range(
    start=series.index[-1] + pd.Timedelta(days=1),
    periods=30,
    freq='D'
)

forecast_df = pd.DataFrame({
    'Date': future_dates,
    'Forecast_Close': forecast_30.values
})

forecast_df
```

Visualisasi Prediksi 30 hari kedepan

```python
# Visualisasi prediksi untuk 30 hari

plt.figure(figsize=(16,7))

# data historis 200 hari terakhir
plt.plot(
    series.index[-200:],
    series[-200:],
    label='Historical Close Price',
    linewidth=2
)

# forecast 30 hari
plt.plot(
    forecast_df['Date'],
    forecast_df['Forecast_Close'],
    label='ARIMA Forecast (30 Days)',
    linewidth=3
)

# titik awal forecast
plt.axvline(
    x=series.index[-1],
    linestyle='--'
)

plt.title('Yahoo Stock Price Forecast - ARIMA Model', fontsize=16)

plt.xlabel('Date', fontsize=12)
plt.ylabel('Close Price', fontsize=12)

plt.legend()

plt.grid(True)

plt.show()
```
