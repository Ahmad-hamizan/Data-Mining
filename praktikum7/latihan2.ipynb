Import Library

```python
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    precision_score,
    recall_score,
    )
```

```python
# Membuat data aktual dan prediksi
aktual = [
    'Lulus',
    'Lulus',
    'Lulus',
    'Lulus',
    'Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Tidak Lulus'
]

prediksi = [
    'Lulus',
    'Lulus',
    'Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Tidak Lulus',
    'Tidak Lulus'
]
```

```python
# Membuat Confusion Matrix
cm = confusion_matrix(
    aktual, 
    prediksi, 
    labels=['Lulus', 'Tidak Lulus']
    )
cm
```

```python
# Menampilkan Confusion Matrix dalam bentuk Tabel
cm_df = pd.DataFrame(
    cm,
    index=[
        'Aktual Lulus', 
        'Aktual Tidak Lulus'
    ],
    columns=[
        'Prediksi Lulus', 
        'Prediksi Tidak Lulus'
    ]
)
cm_df
```

```python
# Menghitung Akurasi
akurasi = accuracy_score(
    aktual, 
    prediksi,
    )

print(f"Akurasi: {akurasi*100:.2f}%")
```

```python
# Menghitung precision
precision = precision_score(
    aktual, 
    prediksi,
    pos_label='Lulus'
)

print(f"Precision: {precision*100:.2f}%")
```

```python
# Menghitung Recall
recall = recall_score(
    aktual, 
    prediksi,
    pos_label='Lulus'
)
print(f"Recall: {recall*100:.2f}%")
```
